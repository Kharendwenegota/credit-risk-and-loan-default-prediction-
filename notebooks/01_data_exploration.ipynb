## 1. Load the data

```python
import pandas as pd

app_train = pd.read_csv('../data/raw/application_train.csv')

print("Shape:", app_train.shape)
app_train.head()
```

## 2. Target distribution
How imbalanced is the default flag (`TARGET`)?

```python
print(app_train['TARGET'].value_counts())
print(app_train['TARGET'].value_counts(normalize=True) * 100)
```

## 3. Missing values
Columns with missing data, sorted by percentage (top 20).

```python
missing = app_train.isnull().sum()
missing_pct = (missing / len(app_train)) * 100
missing_df = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
missing_df = missing_df[missing_df['missing_count'] > 0].sort_values('missing_pct', ascending=False)
missing_df.head(20)
```

## 4. Data types

```python
app_train.dtypes.value_counts()
```

## 5. Column descriptions
Official description of each column in the application table.

```python
col_desc = pd.read_csv('../data/raw/HomeCredit_columns_description.csv', encoding='latin1')
col_desc[col_desc['Table'] == 'application_{train|test}.csv']
```

## 6. Categorical columns
Unique values and counts for the first five text columns.

```python
categorical_cols = app_train.select_dtypes(include='object').columns.tolist()
print(categorical_cols)
for col in categorical_cols[:5]:
    print(f"\n{col}:")
    print(app_train[col].value_counts())
```

## 7. Key numeric columns
Summary statistics for income, credit, annuity, age and employment. The `DAYS_*` columns are day counts relative to the application date.

```python
app_train[['AMT_INCOME_TOTAL', 'AMT_CREDIT', 'AMT_ANNUITY', 'DAYS_BIRTH', 'DAYS_EMPLOYED']].describe()
```
